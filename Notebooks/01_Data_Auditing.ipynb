```python
year_summary = (
    df.groupby("Year")
      .agg(
          Total_Institutions=("Institute_ID", "count"),
          Unique_Institutions=("Institute_ID", "nunique"),
          Categories=("Category", "nunique")
      )
      .reset_index()
)

display(year_summary)
```

```python
coverage = pd.crosstab(
    df["Year"],
    df["Category"]
)

display(coverage)
```

```python
missing_rows = df[df.isnull().any(axis=1)]

print("Number of rows containing missing values:", len(missing_rows))

display(
    missing_rows[
        ["Year", "Category", "Institute_ID", "Name",
         "Score", "Rank", "TLR", "RPC", "GO", "OI", "Perception"]
    ].head(20)
)
```

```python
print("Score data type:", df["Score"].dtype)

print("\nMost common Score values:")
display(df["Score"].value_counts(dropna=False).head(20))

print("\nExamples of Score values:")
display(df["Score"].dropna().unique()[:30])
```

```python
print("Rank data type:", df["Rank"].dtype)

print("\nMost common Rank values:")
display(df["Rank"].value_counts(dropna=False).head(30))

print("\nNumber of unique Rank values:", df["Rank"].nunique())
```

```python
# Identify Score values that cannot be converted to numbers
score_numeric = pd.to_numeric(df["Score"], errors="coerce")

special_scores = df[
    df["Score"].notna() & score_numeric.isna()
]

print("Non-numeric Score records:")
display(
    special_scores[
        ["Year", "Category", "Institute_ID", "Name", "Score", "Rank"]
    ]
)

# Identify Rank values that cannot be converted to numbers
rank_numeric = pd.to_numeric(df["Rank"], errors="coerce")

special_ranks = df[
    df["Rank"].notna() & rank_numeric.isna()
]

print("Non-numeric Rank records:")
display(
    special_ranks[
        ["Year", "Category", "Institute_ID", "Name", "Rank", "Score"]
    ]
)
```

```python
category_quality = (
    df.groupby(["Year", "Category"])
      .agg(
          Total_Records=("Institute_ID", "count"),
          Missing_Score=("Score", lambda x: x.isna().sum()),
          Missing_TLR=("TLR", lambda x: x.isna().sum()),
          Missing_RPC=("RPC", lambda x: x.isna().sum()),
          Missing_GO=("GO", lambda x: x.isna().sum()),
          Missing_OI=("OI", lambda x: x.isna().sum()),
          Missing_Perception=("Perception", lambda x: x.isna().sum())
      )
      .reset_index()
)

display(
    category_quality[
        category_quality[
            ["Missing_Score", "Missing_TLR", "Missing_RPC",
             "Missing_GO", "Missing_OI", "Missing_Perception"]
        ].sum(axis=1) > 0
    ]
)
```
